# Deployment and Remote Testing

This document describes how the machine learning inference service and web client are deployed and accessed remotely. All services run on a virtual machine (VM), while users and developers interact with the system from their local laptop.

---

## Connecting to the Virtual Machine

To deploy and operate the system, developers connect to the virtual machine remotely using Secure Shell (SSH).

### Network Access

Access to the virtual machine requires connection to the university network. This can be achieved by being physically on campus or by connecting through the university VPN. Without this network access, the VM cannot be reached.

### SSH Connection

Once connected to the correct network, the VM is accessed from a terminal on the local machine:

```bash
ssh ubuntu@<VM-IP>
````

Here, `<VM-IP>` is the internal IP address assigned to the virtual machine. Authentication is performed using SSH key-based login.

Example
```
ssh ubuntu@172.27.62.131
````

After successful authentication, a terminal session on the VM is opened, allowing remote execution of commands.

---

## FastAPI Inference Service

The trained regression model is exposed through a FastAPI service. This service receives a record of feature values in JSON format and returns a predicted expert score.

### Setup and Startup

```bash
cd machine_learning_project_4dt907/ml/src/api
python3 -m venv venv
source venv/bin/activate
pip install fastapi uvicorn
uvicorn main:app --host 127.0.0.1 --port 8000
```

The FastAPI service runs locally on the VM and listens on port 8000.
Interactive API documentation is available at:

```text
http://<VM-IP>:8000/docs
```

---

## Web Client and Backend Server

A Node.js server is used to serve the web client and forward prediction requests to the FastAPI service.

### Setup and Startup

```bash
cd machine_learning_project_4dt907/sw/backend
npm install
npm run dev
```

The Node.js server listens on port 3000 and:

* Serves a web-based user interface
* Forwards user-submitted feature records to the FastAPI inference service
* Returns the predicted expert score to the client

---

## Reverse Proxy and Public Access

To allow remote access using a standard web browser, an Nginx reverse proxy is used.

* Incoming HTTP traffic on port 80 is forwarded to the Node.js server on port 3000
* The Node.js server communicates internally with FastAPI on port 8000

This setup provides a single public entry point while keeping internal services isolated.

---

## Remote Testing

Once all services are running, the application can be accessed remotely from a laptop browser by navigating to:

```text
http://<VM-IP>/
```

The web interface allows users to submit a JSON record of feature values. The request is processed by the backend server, forwarded to the FastAPI inference service, and the predicted expert score is returned and displayed in the browser.
