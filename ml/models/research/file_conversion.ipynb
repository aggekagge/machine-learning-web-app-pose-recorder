```python
import pandas as pd
import numpy as np

def normalize_decimal_commas(df: pd.DataFrame) -> pd.DataFrame:
    # Convert strings like "0,6123" -> "0.6123" then to numeric where possible
    out = df.copy()
    for col in out.columns:
        if out[col].dtype == object:
            s = out[col].astype(str)
            # only replace comma if it looks like a decimal comma (digit,comma,digit)
            s2 = s.str.replace(r"(?<=\d),(?=\d)", ".", regex=True)
            out[col] = s2
            # attempt numeric conversion
            out[col] = pd.to_numeric(out[col], errors="ignore")
    return out
```

```python
# scores_and_weak_links
p1 = OUT_DIR / "20190108_scores_and_weak_links.csv"
df1 = pd.read_csv(p1)
df1 = normalize_decimal_commas(df1)
df1.to_csv(p1, index=False)

# big scores
p2 = OUT_DIR / "AimoScore_WeakLink_big_scores.csv"
df2 = pd.read_csv(p2)
df2 = normalize_decimal_commas(df2)
df2.to_csv(p2, index=False)
```
