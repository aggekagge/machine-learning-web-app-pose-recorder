## Sprint organization
- **Sprint length:** 1 week (as defined by the teacher).
- At the **start of each sprint**:
  - Sprint goals are defined (by the teacher).
  - Goals are broken down into concrete tasks/issues.
- During the sprint:
  - Responsibilities are assigned among group members.
  - SCRUM meetings are held at least three times per week, either in person or via Slack depending on availability.
  - The meetings will concern sprint planning, progress, review and retrospective.
- At the **end of each sprint**:
  - A sprint review is conducted to evaluate results.
  - A sprint retrospective is held to reflect on what worked well and what can be improved.

---

## Git and GitLab workflow
- A **milestone** is created for each sprint.
- Sprint goals are divided into **issues**, which are assigned to team members. Issues are assoicated to its corresponding milestone (i.e. sprint).
- Issues are tracked using the issue board and moved from *Open* to *Closed* as work is completed.
- Development workflow:
  - Each feature/element is developed in a separate branch.
  - A merge request (pull request) is created for review.
  - Code is reviewed by other team members before merging into `main`.
  - A “one element, one person” policy is followed to minimize merge conflicts.

# Software and service installation and setup

## Communication
- **Slack**
  - A dedicated workspace is used for project communication.

---

## Programming environment

### Development tools
- **VS Code** is used as the primary code editor.

### Machine Learning environment
- **Python**
  - All machine learning work is implemented in Python.
  - Virtual environments (`venv`) are used to manage dependencies.
  - Required packages are specified in `requirements.txt` files to ensure reproducibility.
  - Python version - Group members uses their preferred version for now.
  - Linting - flake8

### Service environment (Machine Learning API)

- **FastAPI** is used to expose trained machine learning models as a RESTful service.
- The FastAPI server loads the selected production model at startup and provides endpoints (e.g., `/predict`) for generating predictions.
- Incoming requests are validated using typed schemas (Pydantic) to ensure correct input format and data integrity.
- The service communicates with the web application backend using HTTP and JSON.
- This separation enables independent development, testing, and deployment of the machine learning and web application components.

---

### Web application environment

- **Frontend:** HTML, CSS, and JavaScript.
  - Provides the user interface for collecting input data and displaying prediction results.

- **Backend:**  
  - **Node.js** is used to build and run the application server, with **npm** for dependency management.
  - Handles application logic, and communication with external services.
  - Sends input data to the FastAPI ML service and receives predictions via HTTP requests.
  - Processes and formats prediction results before returning them to the frontend.
